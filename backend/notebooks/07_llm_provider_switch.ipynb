# Phase 3 · Step 1 — LangChain LLM wrapper + provider config switch

Demos `get_llm()` from `backend/src/llm/providers.py`: the unwired
providers reject cleanly, and the wired-up one (OpenAI) makes a real
call. See `learning/05_llm_layer/01_langchain_provider_switch.md` for the
why behind this design.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))

from src.llm import get_llm
from src.llm.providers import SUPPORTED_PROVIDERS

print("Supported providers:", SUPPORTED_PROVIDERS)
```

## Unwired providers reject cleanly

`anthropic`, `vertexai`, and `bedrock` are valid provider names (the switch
knows about all 4), but each one raises `NotImplementedError` until a
later task fills in its branch — this confirms that happens loudly, not
silently.

```python
for provider in ("anthropic", "vertexai", "bedrock"):
    try:
        get_llm(provider)
    except NotImplementedError as e:
        print(f"{provider}: raised NotImplementedError ({e})")
    else:
        raise AssertionError(f"{provider} should not be wired up yet")
```

## Real call: OpenAI

Needs `backend/.env` filled in from `.env.example` (`OPENAI_API_KEY`).
This is a real, billed API call.

```python
llm = get_llm("openai")
response = llm.invoke("In one sentence, what is retrieval-augmented generation?")
print(response.content)
```
